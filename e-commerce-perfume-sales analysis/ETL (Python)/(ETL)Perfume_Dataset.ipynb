As we can see here that, there are many unnecesary extra types which are needed to be organized.

```python
df['type'] = df['type'].str.lower().str.strip()
```

```python
df['type'].value_counts()
```

```python
import numpy as np
conditions = [
    df['type'].str.contains(r'edp|eau de parfum|parfum|elixir', na=False),
    df['type'].str.contains(r'edt|eau de toilette|toilette', na=False),
    df['type'].str.contains(r'cologne', na=False),
    df['type'].str.contains(r'mist', na=False),
    df['type'].str.contains(r'deodorant', na=False),
]

choices = [
    'Eau de Parfum',
    'Eau de Toilette',
    'Cologne',
    'Mist',
    'Deodorant'
]

df['type_clean'] = np.select(conditions, choices, default='Other')
```

```python
df['type_clean'].value_counts()
```

```python
df[df['type_clean'] == 'Other'].head(20)
```

found some of the perfumes that are EDP but still existing inside others category

```python
conditions = [
    df['type'].str.contains(r'edp|eau de parfum|eau de perfume|parfum|perfume|elixir', na=False),
    df['type'].str.contains(r'edt|eau de toilette|toilette', na=False),
    df['type'].str.contains(r'cologne', na=False),
    df['type'].str.contains(r'mist', na=False),
    df['type'].str.contains(r'deodorant', na=False),
    df['type'].str.contains(r'spray|body spray', na=False),
    df['type'].str.contains(r'cream|lotion', na=False),
    df['title'].str.contains(r'cream|lotion', na=False)
]

choices = [
    'Eau de Parfum',
    'Eau de Toilette',
    'Cologne',
    'Mist',
    'Deodorant',
    'Body Spray',
    'Skin Product'
]

df['type_clean'] = np.select(conditions, choices, default='Other')
```

```python
df[df['type_clean'] == 'Other'].head(20)
```

Need to work with "title" column as it holds extra infos about perfume types


```python
df['title'] = df['title'].str.lower().str.strip()
```

```python
# create mask for problematic rows
mask = df['type_clean'] == 'Other'

# fix using title column

# Eau de Parfum
df.loc[
    mask & df['title'].str.contains(r'edp|eau de parfum|parfum|soft - al-rehab eau de natural perfume spray', case=False, na=False),
    'type_clean'
] = 'Eau de Parfum'

# Eau de Toilette
df.loc[
    mask & df['title'].str.contains(r'edt|eau de toilette', case=False, na=False),
    'type_clean'
] = 'Eau de Toilette'

# Cologne
df.loc[
    mask & df['title'].str.contains(r'cologne', case=False, na=False),
    'type_clean'
] = 'Cologne'

# Mist
df.loc[
    mask & df['title'].str.contains(r'mist', case=False, na=False),
    'type_clean'
] = 'Mist'

# Deodorant
df.loc[
    mask & df['title'].str.contains(r'deodorant', case=False, na=False),
    'type_clean'
] = 'Deodorant'
```

```python
df['type_clean'].value_counts()
```

```python
# drop old column
df = df.drop(columns=['type'])

# rename new cleaned column
df = df.rename(columns={'type_clean': 'type'})
```

```python
df.head()
```

```python
df['priceWithCurrency'].unique()
```

cleaning the pricing and currency sections for more clear representation of data. and '/ea' seems like a unit that says ".." $ each unit. so fixing it too

```python
df['currency'] = df['priceWithCurrency'].str.extract(r'(US \$)')
df['unit'] = df['priceWithCurrency'].str.extract(r'(/ea)')
df.drop(columns=['priceWithCurrency'], inplace=True)
```

```python
# Drop derived columns (we'll compute in SQL later)
df = df.drop(columns=['update_month', 'update_hour'], errors='ignore')
```

##**Loading the Dataset**

```python
!pip install google-cloud-bigquery pandas -q

from google.colab import auth
import pandas as pd
from google.cloud import bigquery

# Authenticate
auth.authenticate_user()
# BIGQUERY SETUP

project_id = "perfume-e-commerce-analysis"
dataset_id = "perfume_data"
table_id = "cleaned_perfume_data"

client = bigquery.Client(project=project_id)
table_full_id = f"{project_id}.{dataset_id}.{table_id}"


# DEFINE SCHEMA

job_config = bigquery.LoadJobConfig(
    write_disposition="WRITE_TRUNCATE",
    schema=[
        bigquery.SchemaField("brand", "STRING"),
        bigquery.SchemaField("title", "STRING"),
        bigquery.SchemaField("price", "FLOAT"),
        bigquery.SchemaField("available", "FLOAT"),
        bigquery.SchemaField("availableText", "STRING"),
        bigquery.SchemaField("sold", "FLOAT"),
        bigquery.SchemaField("lastUpdated", "TIMESTAMP"),
        bigquery.SchemaField("itemLocation", "STRING"),
        bigquery.SchemaField("update_date", "DATE"),
        bigquery.SchemaField("type", "STRING"),
    ]
)

# UPLOAD TO BIGQUERY

job = client.load_table_from_dataframe(df, table_full_id, job_config=job_config)

job.result()

print(f"Cleaned data uploaded (month/hour removed) to {table_full_id}")
```
